```python
from scipy.stats import norm

def predicted_unemployment_rate(threshold, date, unemployment, cutoff = 12):
    unemployment_cutoff = unemployment.loc[:date]
    sample_dates = unemployment_cutoff.tail(cutoff)
    
    diff_std = sample_dates.diff().std()
    diff = threshold - unemployment_cutoff.iloc[-1]

    z_score = diff / diff_std

    prob = norm.cdf(z_score)
    
    
    return 1-prob
    
    
    
```

```python
def predicted_unemployment_rate_discrete(threshold, date, unemployment, cutoff = 120):
    """date is inclusive, forcast next non-includedd date from series"""
    unemployment_cutoff = unemployment.loc[:date]
    sample_dates = unemployment_cutoff.tail(cutoff)
    sample_diffs = sample_dates.diff()

    sample_diffs =  sample_diffs['UNRATE'].tail(cutoff-1).values
    last_unemployment =  unemployment_cutoff.iloc[-1] 
    dist = (last_unemployment[-1] + sample_diffs)
    dist.sort()
    print(f'dist: {dist}')
    
    count = (dist > threshold).sum()
    return count / len(dist)

   

```
